```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1, 2, 3, 5, 6, 7, 8, 10])
y = np.array([10, 15, 25, 30, 22, 40, 45, 47])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
x = np.array([2000, 2005, 2010, 2015, 2020, 2025, 2026])
y = np.array([4, 4.8, 15, 24, 39, 45, 47])

plt.plot(y, linestyle="dotted", color='black', marker='o')
plt.title("Singapure's economic growth trent between 2000 and 2026 years.")
plt.xlabel("Y e a r s")
plt.ylabel("Percentage %")
plt.show()
```

```python
x = np.array([10, 20, 30, 40, 50])
y1 = np.array([20, 50, 70, 55, 80])
y2 = np.array([10, 20, 33, 45, 47])

plt.plot(x, y1, label='Line 1', color='red', marker='s')
plt.plot(x, y2, label='Line 2', color='blue', linestyle='--', marker='o')
plt.title('Multiple Lines Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.show()
```

```python
x = np.array([2000, 2010, 2020, 2030, 2040])
y1 = np.array([20, 50, 70, 55, 80])
y2 = np.array([10, 20, 33, 45, 47])

plt.plot(x, y1, label='Apple', color='midnightblue', marker='s')
plt.plot(x, y2, label='Samsung', color='gold', linestyle='--', marker='o')
plt.title('Apple and Samsung stocks growth ')
plt.xlabel('Y e a r s')
plt.ylabel('Percentage %')
plt.legend()
plt.show()
```

# BAR chart with Numpy and Matplotlib

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([5, 9, 1, 7])

plt.bar(x, y, color='orange')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

```python
x = np.array(['Salary', 'Charity', 'Expence', 'Rent', 'rest'])
y = np.array([700, 30, 300, 150, 220])

plt.bar(x, y, color='Blue')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

# Scatter Plot with Numpy and Matplotlib

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color='purple', s=50)
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([3, 5, 3, 4, 5, 8, 7, 7, 10])
plt.hist(data, bins=5, color='skyblue', edgecolor='black')

plt.title('Histogramm Example')
plt.xlabel('X values')
plt.ylabel('Y count')
plt.show()
```

# Subplots with Numpy and Matplotlib

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title('Plot 2')

plt.tight_layout()
plt.show()
```

# Save Figure

```python
x = np.array([1, 2, 3, 4])
y = np.array([4, 8, 6, 10])

plt.plot(x, y, color='green')
plt.title('Save Figure Example')
plt.savefig('my_plot.png') # Saves in current folder
plt.show()
```

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

plt.style.use('_mpl-gallery')

# Make data
X, Y, Z = axes3d.get_test_data(0.05)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_wireframe(X, Y, Z, rstride=20, cstride=10)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
