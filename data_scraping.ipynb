```python
import os, shutil

source_dir = 'content/ (output_dir: name of image category)/folder inside of image category '
target_dir = 'content/drive/MyDrive/FOOD_101_dataset/pilaf'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
  src_file = os.path.join(source_dir, filename)
  dst_file = os.path.join(target_dir, filename)
  shutil.move(src_file, dst_file)
```

```python
#Multikeyword downloading case

keywords = ['Uzbek kebab', 'Shashlik kebab', 'Beef kebab plate', 'Traditional kebab food']
for kw in keywords:
  downloader.download(
      kw,
      limit=100,
      output_dir='content/KEBAB_images',
      adult_filter_off=False,
      force_replace=False,
      timeout=60
  )

```

```python
#Moving multi files to google drive

import os
import shutil

source_base = '/content/KEBAB_images'
target_dir = '/content/drive/MyDrive/FOOD_101_dataset/kebab'

# Google Drive-da maqsad papkani yaratamiz
os.makedirs(target_dir, exist_ok=True)

# Barcha ichki papka va fayllarni tekshirib chiqamiz
for root, dirs, files in os.walk(source_base):
    for file in files:
        if file.lower().endswith(('.png', '.jpg', '.jpeg', '.webp')):
            src_file = os.path.join(root, file)
            dst_file = os.path.join(target_dir, file)

            # Faylni ko'chiramiz
            shutil.move(src_file, dst_file)
print("All picture successfully moved to google drive!")
```
