```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread('Horse_cv.png')

cv2_imshow(image)
print('SHAPE: ', image.shape)
```

```python
print('SHAPE: ', image.shape)
```

```python
new_height = 1048
new_width = 576

image = cv2.resize(image, (new_width, new_height))

print("NEW_SIZE: ", image.shape)
```

```python
cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
# 463 => height
# 703 => width
# 3   => color channels (RGB) Red, Green, Blue (BGR) => Blue, Green, Red
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
cropped = gray[150:420, 50:250]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(cropped, cv2.ROTATE_90_CLOCKWISE)

cv2_imshow(rotated)
```

```python
flipped = cv2.flip(cropped, 1)
cv2_imshow(flipped)
```

```python
flipped = cv2.flip(image, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite('flipped_horse_image.png', flipped)
```
